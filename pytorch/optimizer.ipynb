```python
import torch
import torch.nn as nn

relu = nn.ReLU()   
                   #  -ve ====> 0
                   # +ve =====> +ve

x = torch.tensor([-2., 3.])

print(relu(x))
```

```python
sigmoid = nn.Sigmoid()

                      # value in between 0 and 1
print(sigmoid(torch.tensor([0.])))
```

```python
tanh = nn.Tanh()
                         # valuein between 1 and -1
print(tanh(torch.tensor([1.])))
```

```python
# Raw outputs (logits)
logits = torch.tensor([2.0, 1.0, 0.5])

# Softmax
softmax = nn.Softmax(dim=0)

probabilities = softmax(logits)

print(probabilities)   




# used for the mutliclassess
```
