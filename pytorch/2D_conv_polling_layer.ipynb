```python
import torch 
import torch.nn as nn
```

```python
image = torch.rand(1, 1, 5, 5)
```

```python
conv_layer = nn.Conv2d(in_channels=1,
                        out_channels=1,
                          kernel_size=3,
                            stride=2,
                             padding=0)
```

```python
output = conv_layer(image)
print(image.shape)
print(output.shape)
```
