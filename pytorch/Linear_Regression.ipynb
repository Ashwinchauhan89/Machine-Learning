```python
import torch
import torch.nn as nn
import torch.optim as optim

# Data
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])   

# How many of Hours of the Students are studying

y = torch.tensor([[20.0], [40.0], [60.0], [80.0]])

# How many Score the Student Per hours 

# Model
model = nn.Linear(1, 1)

# Loss and Optimizer
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)


# Training
for epoch in range(1000): 
    y_pred = model(X)
    loss = loss_fn(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# Prediction
with torch.no_grad():
    print(model(torch.tensor([[5.0]])))
```

```python
import matplotlib.pyplot as plt

with torch.no_grad():
    predictions = model(X)

plt.scatter(X.numpy(), y.numpy(), label="Actual Data")
plt.plot(X.numpy(), predictions.numpy(), label="Regression Line")


plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```
