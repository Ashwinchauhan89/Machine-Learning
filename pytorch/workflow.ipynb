```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
```

Purpose
torch → Core PyTorch library
nn → Neural network layers and loss functions
optim → Optimizers like Adam, SGD
DataLoader → Loads data in batches
TensorDataset → Converts tensors into a dataset

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
X_dummy = torch.randn(100, 10)
print(X_dummy)
```

```python
X_dummy.shape
```

```python
y_dummy = torch.randint(0, 2, (100,)).float().unsqueeze(1)
```

```python
train_dataset = TensorDataset(X_dummy[:80], y_dummy[:80])
val_dataset = TensorDataset(X_dummy[80:], y_dummy[80:])
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True
)
```

```python
val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False
)
```
