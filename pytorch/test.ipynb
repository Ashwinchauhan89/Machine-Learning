What is Autograd?

Autograd automatically calculates gradients during backpropagation.

```python
x = torch.tensor(2.0, requires_grad=True)
print(x)
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x**3

y.backward()

print(x.grad)
```
