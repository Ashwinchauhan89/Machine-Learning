```python
import torch

# 1. Setup simple linear data: y = 2x
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

# 2. Initialize a single layer model (1 input -> 1 output)
model = torch.nn.Linear(1, 1)

# 3. Choose Mean Squared Error loss and Stochastic Gradient Descent
criterion = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 4. Minimal Training Loop
for epoch in range(100):
    # Step 1: Forward pass
    predictions = model(X)
    
    # Step 2: Compute loss
    loss = criterion(predictions, y)
    
    # Step 3: Clear old gradients
    optimizer.zero_grad()
    
    # Step 4: Backward pass (compute gradients)
    loss.backward()
    
    # Step 5: Update weights
    optimizer.step()

# Test the trained model
print("Prediction for 5.0 (Should be close to 10):", model(torch.tensor([[5.0]])).item())
```
