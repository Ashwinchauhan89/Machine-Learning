```python
import pandas as pd
```

```python
# label encoding

from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("sample_data.csv")

# copy

df_label = df.copy()


le = LabelEncoder()

df_label['Gender_Encoded'] = le.fit_transform(df_label['Gender'])


df_label['Passed_Encoded'] = le.fit_transform(df_label['Passed'])


print(df_label[['Name','Gender','Gender_Encoded',
                'Passed','Passed_Encoded']])


# | Name  | Gender | Gender_Encoded | Passed | Passed_Encoded |
# | ----- | ------ | -------------- | ------ | -------------- |
# | Aman  | Male   | 1              | Yes    | 1              |
# | Priya | Female | 0              | Yes    | 1              |
# | Rahul | Male   | 1              | No     | 0              |
```

What is One-Hot Encoding?

One-Hot Encoding converts categorical text values into separate binary (0/1 or True/False) columns.

```python
df_encoded = pd.get_dummies(
    df_label,
    columns=['City']
)
```
