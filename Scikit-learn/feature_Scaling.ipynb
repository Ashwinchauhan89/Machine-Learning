```python
# | Method          | Formula               | Range          | Outlier Sensitive |
# | --------------- | --------------------- | -------------- | ----------------- |
# | Min-Max Scaling | (X - Min)/(Max - Min) | [0,1]          | Yes               |
# | Standardization | (X - Mean)/Std        | Mean=0, Std=1  | Moderate          |
# | Robust Scaling  | (X - Median)/IQR      | No fixed range | No                |
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    'StudyHours':[2,4,6,8,10],
    'TestScore':[50,60,70,80,90]
})

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

print(pd.DataFrame(
    scaled_data,
    columns=['StudyHours','TestScore']
))
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    'StudyHours':[2,4,6,8,10],
    'TestScore':[50,60,70,80,90]
})

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)

print(pd.DataFrame(
    scaled_data,
    columns=['StudyHours','TestScore']
))
```
