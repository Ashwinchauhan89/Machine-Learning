```python
# 1. What is Classification?

# Classification is a Supervised Machine Learning technique used to predict categories or classes.
```

```python
# | Input         | Output               |
# | ------------- | -------------------- |
# | Email         | Spam / Not Spam      |
# | Student Marks | Pass / Fail          |
# | Patient Data  | Disease / No Disease |
# | Image         | Cat / Dog            |
```

```python
# Unlike Linear Regression, Classification predicts labels, not continuous numbers.
```

```python
# Types of Classification


# Binary Classification

# Only 2 classes.

# Examples:

# Yes / No
# Pass / Fail
# Spam / Not Spam






# Multi-Class Classification

# More than 2 classes.

# Examples:

# Red, Green, Blue
# Cat, Dog, Horse
```
