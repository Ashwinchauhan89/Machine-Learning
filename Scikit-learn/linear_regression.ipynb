```python
# What is Linear Regression?

# Linear Regression finds a relationship between input (X) and output (y) using a straight line.
```

```python
# y=mx+b
```

```python
# Where:

# y = Predicted value
# x = Input feature
# m = Slope
# b = Intercept
```

```python
# | Study Hours (X) | Marks (y) |
# | --------------- | --------- |
# | 1               | 10        |
# | 2               | 20        |
# | 3               | 30        |
# | 4               | 40        |
# | 5               | 50        |
```

```python
from sklearn.linear_model import LinearRegression

# Input Feature
X = [[1], [2], [3], [4], [5]]

# Target Output
y = [10, 20, 30, 40, 50]

# Create Model
model = LinearRegression()

# Train Model
model.fit(X, y)

# Predict
result = model.predict([[6]])

print(result)
```
