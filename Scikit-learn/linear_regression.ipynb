```python
# What is Linear Regression?

# Linear Regression finds a relationship between input (X) and output (y) using a straight line.
```

```python
# y=mx+b
```

```python
# Where:

# y = Predicted value
# x = Input feature
# m = Slope
# b = Intercept
```

```python
# | Study Hours (X) | Marks (y) |
# | --------------- | --------- |
# | 1               | 10        |
# | 2               | 20        |
# | 3               | 30        |
# | 4               | 40        |
# | 5               | 50        |
```

```python
from sklearn.linear_model import LinearRegression

# Input Feature
X = [[1], [2], [3], [4], [5]]

# Target Output
y = [10, 20, 30, 40, 50]

# Create Model
model = LinearRegression()

# Train Model
model.fit(X, y)

# Predict
result = model.predict([[6]])

print(result)
```

```python
# Dataset
#    ↓
# X (Features)
# y (Target)
#    ↓
# model = LinearRegression()
#    ↓
# model.fit(X, y)
#    ↓
# Model learns pattern
#    ↓
# model.predict(new_data)
#    ↓
# Predicted Output
```

```python
# Interview Questions
# Q1. What does fit() do?

# Answer:
# fit() trains the machine learning model using the provided feature data (X) and target data (y). It learns the relationship between input and output variables.

# Q2. What does predict() do?

# Answer:
# predict() uses the trained model to estimate outputs for new, unseen input data.

# Q3. Why do we use [[6]] instead of [6]?

# Answer:
# Scikit-Learn expects input features in a 2D format (samples, features). Therefore a single value must be written as [[6]], representing one sample with one feature.
```
