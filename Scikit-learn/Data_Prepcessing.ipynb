```python
import pandas as pd
```

```python
data = {
    "name" : ["Ashwin","Kajal","Sagar"],
     "age" : [20,None,123],
    "gender" : ["m","f","m"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df.dropna())
```

```python
df["age"].fillna(df["age"].mean(), inplace=True)
print(df)
```
